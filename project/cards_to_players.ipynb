```python
import cv2
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
import numpy as np

from segmentation import segmentation_utils as segmentation
%matplotlib inline


data_path = 'iapr-26-uno-vision-challenge'
train_path = data_path + '/train_images/'
test_path = data_path + '/test_images/'
```

```python
og = cv2.imread(train_path + 'L1000851.jpg')
img = cv2.cvtColor(og, cv2.COLOR_BGR2RGB)
print(img.shape)
H, W, _ = img.shape
cv2.imshow('image', img)
plt.imshow(img,cmap='gray')
```

```python
card_points = np.array([[500, 1500], [400, 1600], [550, 1000], [1500, 2300], [1600, 2250], [3500, 1500], [3300, 1400], [1850, 1450]])
plt.plot(card_points[:, 0], card_points[:, 1], 'ro', )
plt.xlim(0, W)
plt.ylim(H, 0)
```

```python
cluster_centers, cluster_labels = segmentation.clustering(card_points, eps=600, min_samples=1)

print(cluster_labels)
plt.scatter(card_points[:, 0], card_points[:, 1], c=cluster_labels, cmap='viridis')
plt.xlim(0, W)
plt.ylim(H, 0)
```

```python
player_ids = segmentation.cluster_2_player_mapping(cluster_centers, cluster_labels, W, H)
print(player_ids)
```
